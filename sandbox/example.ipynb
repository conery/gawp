## Sample Data

Looking at one of the data files...

```python
import pandas as pd
from shapely.geometry import Point, LineString

from bokeh.palettes import Sunset10
from bokeh.io import output_notebook, show
from bokeh.plotting import figure
```

Read the CSV files exported from the XLS Files sandbox (`xls_files.ipynb`)
* `measurements.csv` has coordinates of line segments
* `points.csv` has coordinates of loci

Rename the colums with shorter, single-word names, and assign an index based on a column with unique IDs.

```python
mf = pd.read_csv('measurements.csv', names=['X','Y','Name'], index_col='Name', header=0)
```

```python
mf.head()
```

```python
foci = pd.read_csv('points.csv', names=["X","Y","Z","Unit","Category","Collection","Name","Surpass","Time","ID"
], index_col='ID', header=0)
```

```python
foci.head()
```

```python
mf.index
```

```python
mf.loc['A']
```

```python
dct = {}
for i, label in enumerate(mf.index[:-1]):
    head = mf.index[i]
    tail = mf.index[i+1]
    k = head+tail
    p0 = Point(mf.loc[head].X, mf.loc[head].Y)
    p1 = Point(mf.loc[tail].X, mf.loc[tail].Y)
    dct[k] = LineString([p0,p1])
```

```python
dct
```

Distance from foci to first line segment

```python
f0 = Point(foci.loc[625].X, foci.loc[625].Y)
f0.distance(dct['AB'])
```

### Bokeh Plots

```python
output_notebook()
```

Static plot with marker labels

```python
p = figure(title='germline', x_axis_label='x', y_axis_label='y')
p.line(x='X', y='Y', source=mf)
p.scatter(x='X', y='Y', size=5, source=mf)
p.text(x='X', y='Y', text='Name', x_offset=5, y_offset=10, source=mf)
show(p)
```

Plot first few foci and line segments

```python
N = 10    # number of foci to show; use len(foci) to show all of them 

yellow = Sunset10[5]

p = figure(title='germline', x_axis_label='x', y_axis_label='y')
p.line(x='X', y='Y', source=mf)
p.scatter(x='X', y='Y', size=5, source=mf)
p.scatter(x='X', y='Y', size=8, fill_color=yellow, source=foci.head(N))
show(p)
```
